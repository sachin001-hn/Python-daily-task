```python
students = {"Rahul":75,"Amit":32,"Priya":88,"Neha":39,"Rohit":55}
for name,marks in students.items():
    if marks>=40:
        print(name,"Pass")
    elif marks > 60:
        print(name,"Pass")
    else:
        print(name,"Fail")
```

```python
students={"Rahul":85,"Amit":35,"Priya":72,"Neha":48}
for name,marks in students.items():
    if marks>=80:
        print(name,"Excellent")
    elif marks>=60:
        print(name,"Good")
    elif marks>=40:
        print(name,"Average")
    else:
        print(name,"Fail")
```

```python
product = {"Laptop":10,"Mouse":3,"Keyboard":0,"Monitor":7,"Printer":2}
for elec,quan in product.items():
        if quan <= 5:
            print(elec,"Low stock")
        elif quan > 5:
            print(elec,"Available")
else:
    print(elec,"Out of stock")
```

```python
product = {"Laptop":10,"Mouse":3,"Keyboard":0,"Monitor":7,"Printer":2}
for elec,stock in product.items():
    if stock==0:
        print(elec,"Out of stock")
    elif stock<=5:
        print(elec,"Low stock")
    elif stock>5:
        print(elec,"Available")

        
```

```python
data = {"a":10,"b":15,"c":25,"d":30,"e":40}
for name,marks in data.items():
    if marks%2==0:
        if marks >= 20:
            print(name,"large even")
        elif marks <= 20:
            print(name,"even")
    else:
        print(name,"odd")
```

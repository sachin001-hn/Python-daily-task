```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
elif units <= 300:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10
else:
    bill = (100 * 5) + (100 * 7) + (100 * 10) + (units - 300) * 15

# Ab surcharge check karo
if bill > 3000:
    bill = bill + (bill * 0.10)   # 10% surcharge add

print("Total Bill = ₹", bill)
```

```python
experience = float(input("Experience (years): "))
salary = float(input("Salary: "))
performance = float(input("Performance score: "))

if experience >= 1:
    if salary >= 50000:
        if performance >= 90:
            bonus_percent = 20
        elif performance >= 75:
            bonus_percent = 10
        else:
            bonus_percent = 5
    else:  # salary < 50000
        if performance >= 90:
            bonus_percent = 15
        else:
            bonus_percent = 5
    
    bonus = salary * (bonus_percent / 100)
    print("Bonus =", bonus)
else:
    print("Not eligible for bonus (experience < 1 year)")
```

```python
age = int(input("Enter age: "))
learner_license = input("Learner license available? (yes/no): ")
test_score = int(input("Driving test score: "))

if age >= 18:
    if learner_license.lower() == "yes":
        if test_score >= 60:
            print("License Approved")
        else:
            print("Failed test, license not approved")
    else:
        print("First get learner license")
else:
    print("Not eligible - age less than 18")
```

```python
# Exercise 5: Comparison Program

# Define data types
my_list = [1, 2, 2, 3]
my_tuple = (1, 2, 2, 3)
my_set = {1, 2, 2, 3}
my_frozenset = frozenset([1, 2, 2, 3])

# Print details
print("List:", my_list, "-> Mutable, Duplicates allowed, Syntax []")
print("Tuple:", my_tuple, "-> Immutable, Duplicates allowed, Syntax ()")
print("Set:", my_set, "-> Mutable, Duplicates removed, Syntax {}")
print("FrozenSet:", my_frozenset, "-> Immutable, Duplicates removed, Syntax frozenset()")
```

```python
# Pattern Printing
for i in range(5, 0, -1):   # 5 se 1 tak reverse loop
    for j in range(1, i+1): # 1 se i tak print karega
        print(j, end="")    # same line me print
    print()                # next line ke liye
```

```python
#without list comp....
data=[5,8,7,4,3]
for i in data:
    print(i**2)
print(data)
```

```python
#with list compression
data=[5,8,7,4,3]
newData=[i**2 for i in data]
print(newData)
```

```python
data=[23,12,9,7,26,80,44]
newdata = [i for i in data if i%2==0]
print(newdata)
```

```python
Marks=[60,80,90,65,23,87,12,40]
Result=["Pass"if ex>=60 else "Fail"for ex in Marks]
print(Result)
```

```python
names = ["Priya","Rahul","Pawan","John","Suman"]
result = [n.upper() for n in names]
print(result)
```

```python
#sets functions: add(),  update(),  remove(), discard(), pop(), clear(), copy(), union(), intersection(), difference()


#add method : This method adds one element to asset at random position 

data={10,20,30,4,30,20,79,789,77}
data.add(8000)
print(data)

#update : This method is used to add multiple elements  to the set.

data={10,20,30,4,30,20,79,789,77}
data.update([20,50,40,45])
print(data)

#remove: This method removes a specific element. If the element does not exist then it will raise an error. 
data={10,20,30,4,30,20,79,789,77}
data.remove(30)
print(data)

#discard: This method is also removes a element same as remove method . The basic difference is if u try to remove a value which is not available in asset then it will not raises any error.
data={10,20,30,4,30,20,79,789,77}
data.discard(301)
print(data)

#pop method: It removes and returns an element from the set . bcz sets are unordered you should not assume which element will be remove.
data={10,20,30,4,30,20,79,789,77}
data.pop()

#clear method: this method removes all elements and return empty set.
data={10,20,30,4,30,20,79,789,77}
data.clear()
print(data)

#copy: It creats a copy to a set .
data={10,20,30,4,30,20,79,789,77}
data.copy()

#union: It combines all unique elements from two sets. 
data={10,20,30,4,30,20,79,789,77}
newdata={20,50,496,48}
print(data.union(newdata))
print(data|newdata)


#intersection: This gives elements that a common two both sets.


#difference: It returns elements that a present in the first set but not in the second set .
```

```python
# you can also use some methods of list and tuple in a set .
# Python Dictonary: A dictionary is build in datatype in python used to store data inside in keyvalue pairs 
# Characterstics/Features of dictionary: 1- data is stored in the form of keyvalue pairs.


data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(data)
# A key must be unique 
#The second name replaces the first value 
#dictionary is mutable: We can modify it's data by using their key
```

```python
#mutable
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
data["name"]="Sumit"
print(data)
```

```python
#key is must be immutable....it means we can not use as a key like- list, sets
#common valid keys are-integer, float,strings,tuple because they belongs to immutable types

#valid keys
data={1:"Amit","Age":23,2.3:"Version"}
print(data)
```

```python
# empty Dict
data={}
print(data)
data=dict()
print(data)
```

```python
#accessing dict values:
#method1- Using key
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(data["name"])
#method2- Using Get Method
print(data.get("name"))
```

```python
#What is the difference between accessingg value to a get method ?
#the get method is safer when we are not sure weather a key exist or not.


#adding new element into dict.
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
data["course"]="Python"
print(data)
```

```python
#updating values: we can use update method to add or modify multiple key value pairs 
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
data.update({"name":"Ayush","age":19})
print(data)
```

```python
#dict functions and methods 
#len method 
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(len(data)) #It simply returns how many keys are available in the dict

#keys()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(data.keys()) #(this method list for all available keys in a particular dict)

#values()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(data.values())# (this method returns all the values of a particular dict)

#itoms()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print(data.items()) #(this method returns all key value pairs of a dict)

#pop()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
data.pop("age")#(this method removes a specific key)
print(data)

#clear()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
data.clear()
print(data)#(this method remove everything from a dict and it returns an empty dict)

#copy()
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
newData=data.copy()
print(data)
print(newData)#(this method creates a copy of the dict)
```

```python
#membership operators:we can also use membership operatpors in dict same as list and tuple.
data={"name":"Mohit","age":23,"college":"invertis","name":"Amit"}
print("age" in data)
print("course"in data)
```

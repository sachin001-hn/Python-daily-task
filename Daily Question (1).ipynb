```python
#Q1
numbers=[20,55,30,80,45,90,10]
result=[ex for ex in numbers if ex>50]
print(result)
```

```python
#Q2
numbers=[10,25,35,60,45,15,50,70]
result=[ex for ex in numbers if ex>20 and ex<=50]
print(result)
```

```python
#Q3
numbers=[10,15,20,30,45,60,75]
result=[ex for ex in numbers if ex%3==0 and ex%5==0]
print(result)
```

```python
#Q4
names = ["Amit","Rahul","Neha","Rohit","Sneha"]
result=[ex for ex in names if len(ex)<5]
print(result)
```

```python
#Q10
price=[500,1200,800,2000,1500]
result=[ex*0.90if ex>1000 else ex*0.95 for ex in price]
print(result)
```

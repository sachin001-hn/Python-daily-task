```python
#isalpha function: this method checks weather all characters are alphabet or not.

data="ayush"
print(data.isalpha())    #true condition

data="ayush123"
print(data.isalpha())    #false condition
```

```python
#isdigit function : this method checks weather all characters are digits or not.
data="123"
print(data.isdigit())
```

```python
#isalnum : this method checks wheather all characters are alphanets or numbers.
data="ayush123"
print(data.isalnum())  #true condition

data="ayush 123"
print(data.isalnum())  #false condition(due to special character"space")
```

```python
#islower : this method checks all the alphabetic characters are lower case or not.
data="ayush"
print(data.islower())  #true condition

data="Ayush"
print(data.islower())  #false condition
```

```python
#is upper : this method checks wheather all alphabetic characters are uppercase or not.
data="AYUSH"
print(data.isupper())  #true condition

data="AYUSH SAXENa"
print(data.isupper())  #false condition
```

```python
# to count a particular letter in given string
data="Python Programming"
i = 0
count = 0
while(i <len(data)):
    if data[i] == "P":
        count+=1
    i += 1
print(count)
```
